### المشكلة 1 — تحويل الوقت (أرقام: `//` و `%`)
</div>

<div dir="rtl">
لديك 205 دقيقة من المذاكرة. حوّلها إلى ساعات ودقائق: القسمة الصحيحة تكتب ساعات، والباقي يكتب دقائق. أكمل السطرين المعلّمين ثم اطبع الجملة.
</div>

<div dir="rtl">
**تذكير سريع:** `//` القسمة الصحيحة تُسقط الكسر (مثل `10 // 3` = 3)، و `%` باقي القسمة (مثل `10 % 3` = 1).
</div>

````python
total_minutes = 205
hours = total_minutes // 60       # أكمل: القسمة الصحيحة
rest = total_minutes % 60         # أكمل: الباقي
print(f"{total_minutes} minutes = {hours} hours {rest} minutes")
````

**المخرجات المتوقعة:**

```text
205 minutes = 3 hours 25 minutes
```

````python
# تحقق ذاتي
assert hours == 3
assert rest == 25
````

### المشكلة 2 — متوسط درجات الطقس (أعداد عشرية + `round`)
</div>

<div dir="rtl">
درجات ثلاثة أيام: 21.5 و 23.0 و 19.8. احسب المتوسط، ثم قرّبه لرقم عشري واحد بـ `round`، ثم اطبع الجملة.
</div>

<div dir="rtl">
**تذكير سريع:** المتوسط = مجموع القيم ÷ عددها، و `round(x, 1)` تقرّب لرقم عشري واحد (جرّبناها مع `0.1 + 0.2` في الحصة).

````python
t1 = 21.5
t2 = 23.0
t3 = 19.8
average = round((t1 + t2 + t3) / 3, 1)   # أكمل: المتوسط ثم round لرقم عشري واحد
print(f"Average temperature: {average} °C")
````


````python
# تحقق ذاتي
assert average == 21.4
````

### المشكلة 3 — تنظيف اسم دولة (نصوص)
</div>

<div dir="rtl">
اسم دولة بفراغات زائدة وأحرف صغيرة: نظّفه (`strip()` ثم `title()`)، ثم استخرج أول ثلاثة حروف، ثم احسب عدد أحرفه بـ `len()`. اطبع الثلاثة مفصولة بـ ` / `.
</div>

<div dir="rtl">
**تذكير سريع:** `.strip()` يزيل الفراغات من الطرفين، و `.title()` يجعل أول حرف كل كلمة كبيرًا، و `cleaned[0:3]` تأخذ أول ثلاثة أحرف، و `len()` تُرجع عدد الأحرف.

````python
country = " New zealand  "
cleaned = country.strip().title()     # أكمل: strip().title()
first3 = cleaned[0:3]      # أكمل: أول ثلاثة حروف
letters = len(cleaned)     # أكمل: len(cleaned)
print(f" {cleaned} / {first3} / {letters} ")
````

````python
# 1. تعريف المتغيرات وتحديد قيمة الدرجة
grade = 85           # يمكنك تغييرها وتجربة القيم الأخرى: 95، 75، 65، 40، 55
level = None

# 2. كتابة سلسلة الشروط (من الأكبر إلى الأصغر)
if grade >= 90:
    level = "A"
elif grade >= 80:
    level = "B"
elif grade >= 70:
    level = "C"
elif grade >= 60:
    level = "D"
else:
    level = "F"

# 3. طباعة النتيجة
print(f"الدرجة {grade} -> {level}")

# -------------------------------------------------------------
# التحقق الذاتي (تأكيد أن الحل صحيح ويعمل بدون أخطاء)
assert level in ("A", "B", "C", "D", "F")
assert not (grade >= 90 and level != "A"), ">=90: A"
assert not (80 <= grade < 90 and level != "B"), "80-89: B"
assert not (70 <= grade < 80 and level != "C"), "70-79: C"
assert not (60 <= grade < 70 and level != "D"), "60-69: D"
assert not (grade < 60 and level != "F"), "<60: F"

print("A1: ممتاز")
````

````python
import random

# 1. تحديد الرقم العشوائي المستهدف وتهيئة المحاولات
target = random.randint(1, 50)
tries = 0

# 2. استقبال التخمين الأول وزيادة العداد للمحاولة الأولى
guess = int(input("خمن رقم من 1 لـ 50: "))
tries += 1

# 3. حلقة التكرار (while) طالما أن التخمين غير مساوي للهدف
while guess != target:
    # طباعة التوجيه (أكثر أو أقل) بناءً على التخمين
    if guess < target:
        print("أكثر")
    else:
        print("أقل")

    # استقبال تخمين جديد من المستخدم
    guess = int(input("خمن رقم جديد: "))
    # زيادة عداد المحاولات بمقدار 1 في كل مرة يخمّن فيها
    tries += 1

# 4. رسالة الفوز عند الخروج من الحلقة بنجاح
print(f"برافو! عدد المحاولات {tries}")

# -------------------------------------------------------------
# التحقق الذاتي
assert tries >= 1, "لسه صفر - العب اللعبة الأول"
print("A3: ممتاز")
````

````python
def is_even(n):
    # تلميح: رجّع True لو باقي قسمة n على 2 يساوي 0
    return n % 2 == 0


def square(n):
    # تلميح: اضرب n في نفسه ورجّع الناتج
    return n * n


def average(values):
    # تلميح: اقسم مجموع القيم على عددها
    # تأكد إن القائمة مش فاضية عشان ما يحصلش خطأ قسمة على صفر
    if not values:
        return 0
    return sum(values) / len(values)


# سطر الطباعة للتجربة
print(is_even(4), square(7))
````

````python
%%writefile tracker/functions.py
import csv

def add_record(records, record):
    """Append one record dict to the list."""
    records.append(record)
    return records

def filter_records(records, category):
    """Return records matching the given category."""
    return [r for r in records if r["category"] == category]

def summarize(records):
    """Return count, total, average, biggest, smallest of amount_egp."""
    if not records:
        return {"count": 0, "total": 0.0, "average": 0.0, "biggest": 0.0, "smallest": 0.0}

    amounts = [r["amount_egp"] for r in records]
    return {
        "count": len(amounts),
        "total": sum(amounts),
        "average": sum(amounts) / len(amounts),
        "biggest": max(amounts),
        "smallest": min(amounts)
    }

def save_records(records, path):
    """Write records to CSV with a header row."""
    if not records:
        return
    with open(path, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        # كتابة العناوين (المفاتيح من أول قاموس)
        writer.writerow(records[0].keys())
        # كتابة البيانات
        for r in records:
            writer.writerow(r.values())

def load_records(path):
    """Read CSV back into dicts; a missing file returns an empty list."""
    try:
        with open(path, "r", encoding="utf-8") as f:
            reader = csv.DictReader(f)
            loaded = []
            for row in reader:
                # تحويل البيانات الرقمية لأن الـ CSV بيقراها نصوص
                row["amount_egp"] = float(row["amount_egp"])
                row["quantity"] = int(row["quantity"])
                loaded.append(row)
            return loaded
    except FileNotFoundError:
        return []

# -------------------------------------------------------------
# دوال الجزء C (حطيتها هنا عشان تجمع كل دوال المشروع في ملف واحد)
# -------------------------------------------------------------

def delete_record(records, index):
    """Delete the record at index and return it."""
    if index < 0 or index >= len(records):
        return None
    return records.pop(index)

def update_record(records, index, field, value):
    """Change one field of the record at index."""
    if index < 0 or index >= len(records):
        return False
    # تحويل الأنواع لو بنعدل حقول رقمية
    if field == "amount_egp":
        value = float(value)
    elif field == "quantity":
        value = int(value)

    records[index][field] = value
    return True
````

````python
class Student:
    def __init__(self, name, grades):
        # تخزين الاسم والدرجات من المعاملات
        self.name = name
        self.grades = grades

    def average(self):
        # حساب المتوسط بـ for loop بدون sum()
        if not self.grades:
            return 0
        total = 0
        for grade in self.grades:
            total += grade
        return total / len(self.grades)

s = Student("Sara", [85, 90, 76])
print(s.name)
````

````python
class Classroom:
    def __init__(self, room):
        self.room = room
        self.students = []

    def add(self, student):
        self.students.append(student)

    def best(self):
        if not self.students:
            return None
        # افتراض أول طالب هو الأفضل واللف على الباقي للمقارنة
        best_student = self.students[0]
        for s in self.students:
            if s.average() > best_student.average():
                best_student = s
        return best_student

room = Classroom("A101")
room.add(Student("Sara", [85, 90, 76]))
````

````python
%%writefile tracker/models.py
import csv

class Record:
    """One tracker row: the five project fields."""
    def __init__(self, date, category, amount_egp, payment_method, quantity):
        self.date = date
        self.category = category
        self.amount_egp = float(amount_egp) # حفظ المبلغ كرقم عشري
        self.payment_method = payment_method
        self.quantity = int(quantity)       # حفظ الكمية كرقم صحيح

    def to_dict(self):
        """Return records matching the given category."""
        return {
            "date": self.date,
            "category": self.category,
            "amount_egp": self.amount_egp,
            "payment_method": self.payment_method,
            "quantity": self.quantity
        }

    def __str__(self):
        """Return formatted string directly for printing."""
        return f"[{self.date}] {self.category}: {self.amount_egp} EGP ({self.payment_method}) x{self.quantity}"


class Tracker:
    """A list of Record objects + every behavior the menu needs."""
    def __init__(self):
        self.records = [] # قائمة فاضية لتخزين كائنات Record

    def add(self, record):
        """Add a Record object to the list."""
        self.records.append(record)

    def filter_by(self, category):
        """Return a list of records matching the given category."""
        return [r for r in self.records if r.category == category]

    def summary(self):
        """Return count, total, average, biggest, smallest of amount_egp."""
        if not self.records:
            return {"count": 0, "total": 0.0, "average": 0.0, "biggest": 0.0, "smallest": 0.0}

        amounts = [r.amount_egp for r in self.records]
        return {
            "count": len(amounts),
            "total": sum(amounts),
            "average": sum(amounts) / len(amounts),
            "biggest": max(amounts),
            "smallest": min(amounts)
        }

    def save(self, path):
        """Write records to CSV with a header row."""
        if not self.records:
            return
        with open(path, "w", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)
            # صف العناوين أول حاجة
            writer.writerow(["date", "category", "amount_egp", "payment_method", "quantity"])
            # كتابة البيانات
            for r in self.records:
                writer.writerow([r.date, r.category, r.amount_egp, r.payment_method, r.quantity])

    def load(self, path):
        """Read CSV back into Record objects."""
        try:
            with open(path, "r", encoding="utf-8") as f:
                reader = csv.DictReader(f)
                self.records = []
                for row in reader:
                    rec = Record(
                        date=row["date"],
                        category=row["category"],
                        amount_egp=row["amount_egp"],
                        payment_method=row["payment_method"],
                        quantity=row["quantity"]
                    )
                    self.records.append(rec)
        except FileNotFoundError:
            # لو الملف مش موجود بنسيب القائمة زي ما هي
            pass

    # -------------------------------------------------------------
    # دالة الجزء C الإلزامية (توضع داخل كلاس Tracker)
    # -------------------------------------------------------------
    def top_n(self, field, n):
        """Return the n records with the biggest value of field."""
        # فرز السجلات تنازلياً بناءً على الحقل المطلوب باستخدام getattr
        sorted_records = sorted(self.records, key=lambda r: getattr(r, field), reverse=True)
        return sorted_records[:n]
````

````python
from tracker.models import Record, Tracker

t = Tracker()
t.load("data/sample_records.csv")

# لو الملف لسه فاضي، بنلصق السجلات ككائنات Record (لازم 10 على الأقل)
if not t.records:
    records_data = [
        ["2026-10-01", "food", 120.0, "cash", 1],
        ["2026-10-02", "drink", 45.0, "card", 2],
        ["2026-10-03", "food", 85.5, "cash", 1],
        ["2026-10-04", "transport", 50.0, "cash", 1],
        ["2026-10-05", "clothes", 450.0, "card", 1],
        ["2026-10-06", "drink", 30.0, "cash", 1],
        ["2026-10-07", "food", 200.0, "card", 1],
        ["2026-10-08", "entertainment", 150.0, "cash", 3],
        ["2026-10-09", "transport", 40.0, "cash", 1],
        ["2026-10-10", "food", 95.0, "cash", 1]
    ]
    for r in records_data:
        t.add(Record(r[0], r[1], r[2], r[3], r[4]))
    t.save("data/sample_records.csv")

print(f"عدد السجلات ككائنات: {len(t.records)}")
````

````python
while True:
    print("\n--- قائمة التحكم بالـ Tracker (كائنات V3) ---")
    print("1) عرض الكل")
    print("2) إضافة سجل")
    print("3) تصفية حسب التصنيف")
    print("4) ملخص")
    print("5) بحث بالكلمة")
    print("6) حذف سجل")
    print("7) تعديل سجل")
    print("8) خروج وحفظ")
    choice = input("اختار: ")

    if choice == "1":
        for r in t.records:
            print(r)  # هيطبع باستخدام دالة __str__ المنظمة

    elif choice == "2":
        try:
            date = input("التاريخ (YYYY-MM-DD): ")
            category = input("التصنيف: ")
            amount = float(input("المبلغ بالجنيه: "))
            method = input("طريقة الدفع: ")
            qty = int(input("الكمية: "))

            new_rec = Record(date, category, amount, method, qty)
            t.add(new_rec)
            print("ممتاز! تم إضافة الكائن بنجاح.")
        except ValueError:
            print("خطأ: يرجى إدخال أرقام صحيحة في خانة المبلغ والكمية!")

    elif choice == "3":
        category = input("التصنيف: ")
        for r in t.filter_by(category):
            print(r)

    elif choice == "4":
        print(t.summary())

    elif choice == "5":
        word = input("كلمة: ")
        for r in t.records:
            if word in str(r):
                print(r)

    elif choice == "6":
        for idx, r in enumerate(t.records):
            print(f"{idx}) {r}")
        try:
            del_idx = int(input("أدخل رقم السجل للحذف: "))
            if 0 <= del_idx < len(t.records):
                deleted = t.records.pop(del_idx)
                print(f"تم حذف: {deleted}")
            else:
                print("رقم السجل غير موجود.")
        except ValueError:
            print("يرجى إدخال رقم صحيح.")

    elif choice == "7":
        for idx, r in enumerate(t.records):
            print(f"{idx}) {r}")
        try:
            up_idx = int(input("أدخل رقم السجل للتعديل: "))
            if 0 <= up_idx < len(t.records):
                field = input("اسم الحقل (date, category, amount_egp, payment_method, quantity): ")
                value = input("القيمة الجديدة: ")

                if field == "amount_egp":
                    setattr(t.records[up_idx], field, float(value))
                elif field == "quantity":
                    setattr(t.records[up_idx], field, int(value))
                else:
                    setattr(t.records[up_idx], field, value)
                print("تم التعديل بنجاح!")
            else:
                print("رقم السجل غير موجود.")
        except ValueError:
            print("مدخلات خاطئة.")

    elif choice == "8":
        t.save("data/sample_records.csv")
        print("اتحفظت! مع السلامة!")
        break
    else:
        print("اختيار غلط — جرّب تاني")
````

````python
assert isinstance(t, Tracker), "t نئاك Tracker"
assert len(t.records) >= 10, "records"
assert all(isinstance(r, Record) for r in t.records), "لجس لك Record"
s = t.summary()
assert s["count"] == len(t.records), "count"
assert s["smallest"] <= s["average"] <= s["biggest"], "bounds"

t.save("data/sample_records.csv")
t2 = Tracker()
t2.load("data/sample_records.csv")

assert len(t2.records) == len(t.records), "roundtrip count"
assert isinstance(t2.records[0], Record), "roundtrip Record"
assert abs(t2.records[0].amount_egp - t.records[0].amount_egp) < 0.01, "roundtrip value"
print("Part B: ممتاز")
````
