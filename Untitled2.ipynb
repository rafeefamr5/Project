```python
# ==========================================
# A1 - سلة المشتريات
# ==========================================
cart = []
cart.append("Latte")
cart.append("Croissant")
cart.append("Tea")

cart.remove("Croissant")
cart.insert(0, "Water")
cart.pop()
cart.sort()

print(cart, len(cart))

# تحقق ذاتي A1
assert cart == ["Latte", "Water"]
assert len(cart) == 2
print("ممتاز A1")


# ==========================================
# A2 - إزالة التكرار باستخدام set
# ==========================================
numbers = [4, 8, 4, 2, 8, 9, 2, 1, 9, 9]
unique = list(set(numbers))
unique.sort()  # ترتيب القائمة

print(unique)

# تحقق ذاتي A2
assert sorted(unique) == [1, 2, 4, 8, 9]
print("ممتاز A2")


# ==========================================
# A3 - دفتر الدرجات
# ==========================================
grades = {"math": 88, "english": 76}
grades["math"] = 92          # تعديل قيمة
grades["physics"] = 79       # إضافة مفتاح جديد
del grades["english"]        # حذف مفتاح

best = max(grades.values())  # أعلى درجة
print(grades, best)

# تحقق ذاتي A3
assert grades == {"math": 92, "physics": 79}
assert best == 92
print("ممتاز A3")


# ==========================================
# A4 - فك الـ tuple (unpacking)
# ==========================================
student = (92.5, 21, "ليلى")
score, age, name = student  # فك عناصر الـ tuple

print(name, age, score)

pairs = [("أحمد", 77), ("سارة", 85)]
first_name, first_score = pairs[1]  # تفكيك زوج "سارة" ليمر الاختبار

# تحقق ذاتي A4
assert name == "ليلى" and age == 21 and score == 92.5
assert first_name == "سارة" and first_score == 85
print("ممتاز A4")


# ==========================================
# A5 - إصلاح فخ التشابه (aliasing)
# ==========================================
prices_a = [10, 20, 30]
prices_b = prices_a.copy()  # استخدام copy() لعمل نسخة مستقلة

prices_b.append(40)
print(prices_a)

# تحقق ذاتي A5
assert prices_a == [10, 20, 30]
assert prices_b == [10, 20, 30, 40]
print("ممتاز A5")
```
